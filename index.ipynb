```python
# โปรแกรมคำนวณค่าขนส่งตู้คอนเทนเนอร์ (20 ฟุต และ 40 ฟุต) 77 จังหวัด
# อ้างอิงระยะทางโดยประมาณจากศูนย์กลางการขนส่ง (กรุงเทพฯ / ท่าเรือแหลมฉบัง)

province_distances = {
    # ภาคกลาง
    "กรุงเทพมหานคร": 100,
    "สมุทรปราการ": 120,
    "นนทบุรี": 110,
    "ปทุมธานี": 130,
    "พระนครศรีอยุธยา": 160,
    "อ่างทอง": 180,
    "ลพบุรี": 200,
    "สิงห์บุรี": 190,
    "ชัยนาท": 220,
    "สระบุรี": 150,
    "นครปฐม": 140,
    "สมุทรสาคร": 130,
    "สมุทรสงคราม": 150,
    "สุพรรณบุรี": 180,

    # ภาคเหนือ
    "เชียงใหม่": 700,
    "เชียงราย": 800,
    "ลำปาง": 600,
    "ลำพูน": 680,
    "อุตรดิตถ์": 500,
    "แพร่": 550,
    "น่าน": 670,
    "พะเยา": 740,
    "แม่ฮ่องสอน": 850,
    "ตาก": 500,
    "พิษณุโลก": 400,
    "สุโขทัย": 450,
    "เพชรบูรณ์": 350,
    "พิจิตร": 360,
    "กำแพงเพชร": 380,

    # ภาคตะวันออกเฉียงเหนือ (อีสาน)
    "นครราชสีมา": 250,
    "ขอนแก่น": 450,
    "อุดรธานี": 560,
    "อุบลราชธานี": 600,
    "บุรีรัมย์": 350,
    "สุรินทร์": 400,
    "ศรีสะเกษ": 500,
    "ยโสธร": 530,
    "ชัยภูมิ": 330,
    "อำนาจเจริญ": 580,
    "หนองบัวลำภู": 520,
    "หนองคาย": 610,
    "มหาสารคาม": 440,
    "ร้อยเอ็ด": 470,
    "กาฬสินธุ์": 500,
    "สกลนคร": 600,
    "นครพนม": 700,
    "มุกดาหาร": 620,
    "เลย": 540,

    # ภาคใต้
    "ชุมพร": 480,
    "สุราษฎร์ธานี": 650,
    "นครศรีธรรมราช": 780,
    "สงขลา": 950,
    "ภูเก็ต": 860,
    "ระนอง": 570,
    "พังงา": 780,
    "กระบี่": 790,
    "ตรัง": 850,
    "พัทลุง": 880,
    "สตูล": 970,
    "ปัตตานี": 1050,
    "ยะลา": 1080,
    "นราธิวาส": 1150,

    # ภาคตะวันออก
    "ชลบุรี": 30,
    "ระยอง": 90,
    "จันทบุรี": 200,
    "ตราด": 300,
    "ฉะเชิงเทรา": 80,
    "ปราจีนบุรี": 130,
    "สระแก้ว": 220,

    # ภาคตะวันตก
    "กาญจนบุรี": 230,
    "ราชบุรี": 160,
    "เพชรบุรี": 200,
    "ประจวบคีรีขันธ์": 320
}

def calculate_shipping_cost():
    print("=== โปรแกรมคำนวณค่าขนส่งตู้คอนเทนเนอร์ (77 จังหวัด) ===")

    # 1. รับค่าจังหวัดปลายทาง
    province = input("กรุณาใส่ชื่อจังหวัดปลายทาง: ").strip()

    if province not in province_distances:
        print(f"\n[แจ้งเตือน] ไม่พบข้อมูลระยะทางสำหรับจังหวัด '{province}' ในระบบ")
        try:
            custom_distance = float(input("กรุณาระบุระยะทางโดยประมาณ (กม.) จากท่าเรือแหลมฉบัง/กรุงเทพฯ ไปยังจังหวัดนี้: "))
            distance = custom_distance
        except ValueError:
            print("ข้อมูลระยะทางไม่ถูกต้อง กำหนดค่าเริ่มต้นที่ 300 กม.")
            distance = 300
    else:
        distance = province_distances[province]
        print(f"ระยะทางอ้างอิงไปยัง {province}: {distance} กม.")

    # 2. รับขนาดตู้คอนเทนเนอร์
    print("\nขนาดตู้คอนเทนเนอร์ที่มี:")
    print("1. ขนาด 20 ฟุต")
    print("2. ขนาด 40 ฟุต")

    container_choice = input("เลือกขนาดตู้ (พิมพ์ 20 หรือ 40): ").strip()

    if container_choice == "20":
        container_size = "20 ฟุต"
        base_rate = 3500
        rate_per_km = 35
    elif container_choice == "40":
        container_size = "40 ฟุต"
        base_rate = 5000
        rate_per_km = 55
    else:
        print("เลือกขนาดตู้ไม่ถูกต้อง ระบบจะใช้ขนาด 20 ฟุตเป็นค่าเริ่มต้น")
        container_size = "20 ฟุต"
        base_rate = 3500
        rate_per_km = 35

    # 3. รับจำนวนตู้
    try:
        quantity = int(input("กรุณาใส่จำนวนตู้ที่ต้องการขนส่ง (ตู้): "))
        if quantity <= 0:
            quantity = 1
    except ValueError:
        print("จำนวนตู้ไม่ถูกต้อง กำหนดเป็น 1 ตู้")
        quantity = 1

    # 4. คำนวณค่าขนส่ง
    cost_per_unit = base_rate + (distance * rate_per_km)

    discount = 0.0
    if quantity >= 3:
        discount = 0.05
        print("ได้รับส่วนลดพิเศษสำหรับการขนส่งจำนวนมาก (5%)")

    total_cost = (cost_per_unit * quantity) * (1 - discount)

    # 5. แสดงผลลัพธ์ (แก้ไขเครื่องหมายและเพิ่มตัว f หน้า string แล้ว)
    print("\n" + "="*40)
    print("สรุปใบเสนอราคาค่าขนส่ง")
    print("="*40)
    print(f"- จังหวัดปลายทาง: {province}")
    print(f"- ขนาดตู้คอนเทนเนอร์: {container_size}")
    print(f"- จำนวนตู้: {quantity} ตู้")
    print(f"- ระยะทางโดยประมาณ: {distance} กม.")
    print(f"- ค่าขนส่งต่อตู้: {cost_per_unit:,.2f} บาท")
    if discount > 0:
        print(f"- ส่วนลด ({int(discount*100)}%): -{(cost_per_unit * quantity * discount):,.2f} บาท")
    print(f"รวมค่าขนส่งสุทธิ: {total_cost:,.2f} บาท")
    print("="*40)

if __name__ == "__main__":
    calculate_shipping_cost()
```
